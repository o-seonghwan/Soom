## 2. 전처리 및 파생변수
## 3. 기초 통계 및 시각화
## 4. 결과 저장 및 검증

```python

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATASET_KEY = "local_resd_250m_20261028"
RAW_FILENAME = "250_LOCAL_RESD_20261028.csv"

def find_workspace(start=None):
    here = Path.cwd() if start is None else Path(start)
    for p in [here, *here.parents]:
        if (p / "do_not_lung").exists() and (p / "data").exists():
            return p
        if p.name == "do_not_lung" and (p.parent / "data").exists():
            return p.parent
    raise FileNotFoundError("do_not_lung/와 data/ 폴더를 찾지 못했습니다.")

WORKSPACE = find_workspace()
RAW_DIR = WORKSPACE / "data" / "raw_data"
OUT_DIR = WORKSPACE / "data" / "processed_data" / DATASET_KEY
OUT_DIR.mkdir(parents=True, exist_ok=True)

raw = pd.read_csv(
    RAW_DIR / RAW_FILENAME,
    encoding="cp949",
    dtype={"행정동코드": "string", "250M격자": "string"}
)
print("원본 크기:", raw.shape)
print("자료형 및 메모리 사용량:")
display(raw.dtypes.astype(str).to_frame("자료형"))
print(f"{raw.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
display(raw.head())

id_cols = ["일자", "시간", "행정동코드", "250M격자"]
value_cols = [c for c in raw.columns if c not in id_cols]
clean = raw.copy()

clean["일자"] = pd.to_datetime(clean["일자"].astype(str), format="%Y%m%d", errors="coerce")
clean["시간"] = pd.to_numeric(clean["시간"], errors="coerce").astype("Int64")
clean["행정동코드"] = clean["행정동코드"].str.strip()
clean["250M격자"] = clean["250M격자"].str.strip()

# '*'는 0으로 간주하지 않고 결측 처리하며 원본 표기 여부를 보존
for col in value_cols:
    original = clean[col].astype("string").str.strip()
    clean[f"{col}_미제공표시"] = original.eq("*").fillna(False)
    clean[col] = pd.to_numeric(original.mask(original.eq("*")), errors="coerce")

clean["일자_검증"] = clean["일자"].notna()
clean["시간_범위검증"] = clean["시간"].between(0, 23).fillna(False)
clean["행정동코드_길이검증"] = clean["행정동코드"].str.fullmatch(r"\d{8}", na=False)
clean["격자_형식검증"] = clean["250M격자"].str.match(r"^[가-힣][0-9]+$", na=False)

key_cols = ["일자", "시간", "행정동코드", "250M격자"]
exact_duplicates = int(clean.duplicated().sum())
key_duplicates = int(clean.duplicated(key_cols).sum())

male_cols = [c for c in value_cols if c.startswith("남자 ")]
female_cols = [c for c in value_cols if c.startswith("여자 ")]

features = clean[key_cols + ["생활인구합계"]].copy()
features["남성_연령합계"] = clean[male_cols].sum(axis=1, min_count=1)
features["여성_연령합계"] = clean[female_cols].sum(axis=1, min_count=1)
features["연령대별합계_합산"] = features[["남성_연령합계", "여성_연령합계"]].sum(axis=1, min_count=1)
features["남성비율_후보"] = features["남성_연령합계"] / features["연령대별합계_합산"].replace(0, np.nan)
features["여성비율_후보"] = features["여성_연령합계"] / features["연령대별합계_합산"].replace(0, np.nan)

for sex, cols in [("남성", male_cols), ("여성", female_cols)]:
    for col in cols:
        features[f"{sex}_{col.split(' ', 1)[1]}_인구"] = clean[col]

features["생활인구합계_대비_연령합계차이"] = (
    features["생활인구합계"] - features["연령대별합계_합산"]
)
features["시간대"] = clean["시간"].map(
    lambda h: f"{int(h):02d}시" if pd.notna(h) else pd.NA
)
features["주간야간_구분"] = clean["시간"].map(
    lambda h: "주간(06~17시)" if pd.notna(h) and 6 <= int(h) < 18
    else ("야간(18~05시)" if pd.notna(h) else pd.NA)
)
features["생활인구합계_미제공"] = clean["생활인구합계"].isna()

def grouped_summary(frame, by, group_name):
    result = frame.groupby(by, dropna=False)["생활인구합계"].agg(
        관측행수="count", 평균="mean", 중앙값="median", 합계="sum", 최댓값="max"
    ).reset_index()
    result.insert(0, "집계유형", group_name)
    result.rename(columns={by: "그룹값"}, inplace=True)
    return result

summary_parts = [
    grouped_summary(clean, "시간", "시간별"),
    grouped_summary(clean, "행정동코드", "행정동별"),
    grouped_summary(clean, "250M격자", "250m격자별")
]
summary = pd.concat(summary_parts, ignore_index=True)

quality = pd.DataFrame({
    "항목": [
        "원본_행수", "원본_열수", "완전중복행수", "복합키중복행수",
        "고유격자수", "고유행정동수", "고유일자수", "고유시간수",
        "별표표시셀수", "생활인구합계_결측수"
    ],
    "값": [
        len(raw), raw.shape[1], exact_duplicates, key_duplicates,
        clean["250M격자"].nunique(), clean["행정동코드"].nunique(),
        clean["일자"].nunique(), clean["시간"].nunique(),
        int(raw[value_cols].astype("string").eq("*").sum().sum()),
        int(clean["생활인구합계"].isna().sum())
    ]
})
quality.insert(0, "집계유형", "품질요약")
summary = pd.concat([summary, quality], ignore_index=True, sort=False)

# 시간대별 평균 생활인구
plt.figure(figsize=(12, 4))
clean.groupby("시간")["생활인구합계"].mean().plot(kind="bar")
plt.title("2026-10-28 시간대별 250m 격자 생활인구합계 평균")
plt.xlabel("시간"); plt.ylabel("평균 생활인구합계")
plt.tight_layout(); plt.show()

# 생활인구 합계 분포
plt.figure(figsize=(8, 4))
clean["생활인구합계"].dropna().plot(kind="hist", bins=50)
plt.title("생활인구합계 분포")
plt.xlabel("생활인구합계"); plt.ylabel("빈도")
plt.tight_layout(); plt.show()

# 성별 연령대 합계 비교
plt.figure(figsize=(8, 4))
features[["남성_연령합계", "여성_연령합계"]].sum().plot(kind="bar")
plt.title("성별 연령대 합계(관측값 기준)")
plt.ylabel("합계"); plt.tight_layout(); plt.show()

for filename, frame in [
    ("clean.csv", clean), ("summary.csv", summary), ("features.csv", features)
]:
    frame.to_csv(OUT_DIR / filename, index=False, encoding="utf-8-sig")

def schema_lines(frame):
    return "\n".join(
        f"- `{col}` ({dtype})"
        for col, dtype in zip(frame.columns, frame.dtypes.astype(str))
    )

readme = f'''# {DATASET_KEY}

## 데이터 개요
- 원본 파일: `{RAW_FILENAME}`
- 원본 인코딩: CP949
- 원본 크기: {raw.shape[0]:,}행 x {raw.shape[1]}열
- 기준일자: {clean["일자"].min()} (단일 일자)
- 시간대: 데이터에 기록된 시간 컬럼 기준
- 공간 단위: 250m 격자 코드 및 행정동 코드

## 해석 및 사용 시 주의사항
- 원본 수치 컬럼의 `*`는 결측으로 변환하고 0으로 대체하지 않음.
- 각 수치 컬럼의 `*_미제공표시`에 원본의 `*` 표기 여부를 보존함.
- `*`의 정확한 의미(0/비공개/미집계/미제공)는 원본만으로 확정할 수 없음.
- 생활인구는 흡연 발생 또는 간접흡연 노출의 정답 라벨이 아님.
- 단일 날짜 자료이므로 계절·장기 추세로 일반화하지 않음.
- 격자 코드는 식별자이며 이 파일만으로 위도·경도를 직접 계산하지 않음.
- 공통 250m 격자 및 도보 링크의 정의가 확인되기 전까지 다른 공간 데이터와 임의 결합하지 않음.

## 파일 구성
- `clean.csv`: 원본 컬럼 보존, 날짜·시간·코드·수치형 표준화 및 검증 플래그 추가.
- `summary.csv`: 시간별·행정동별·250m격자별 생활인구 통계와 품질 요약.
- `features.csv`: 성별·연령대 합계, 성별 비율 후보, 시간대 및 주야간 구분 등 분석 변수 후보.
- `clean.csv`와 `features.csv`는 `일자`, `시간`, `행정동코드`, `250M격자` 복합키로 연결.
- `summary.csv`는 그룹 집계 결과이므로 원본 행 단위 테이블과 1:1 연결되지 않음.

## clean.csv 컬럼 및 자료형
{schema_lines(clean)}

## summary.csv 컬럼 및 자료형
{schema_lines(summary)}

## features.csv 컬럼 및 자료형
{schema_lines(features)}
'''
(OUT_DIR / "README.md").write_text(readme, encoding="utf-8")

# 저장 후 재읽기 및 스키마/인코딩 검증
for filename, expected in [
    ("clean.csv", clean), ("summary.csv", summary), ("features.csv", features)
]:
    check = pd.read_csv(OUT_DIR / filename, encoding="utf-8-sig")
    assert check.shape == expected.shape
    assert list(check.columns) == list(expected.columns)
assert all(f"`{col}`" in readme for col in clean.columns)
assert all(f"`{col}`" in readme for col in summary.columns)
assert all(f"`{col}`" in readme for col in features.columns)

print("전처리 후 크기:", clean.shape)
print("summary 크기:", summary.shape)
print("features 크기:", features.shape)
print("완전중복:", exact_duplicates, "복합키중복:", key_duplicates)
print("별표 표시 셀:", int(raw[value_cols].astype("string").eq("*").sum().sum()))
print("CSV 재읽기 및 README 스키마 검증 완료")
```
